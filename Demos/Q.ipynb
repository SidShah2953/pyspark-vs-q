# Demo 1

```q
// Generate 10 million rows of simulated market data
n:10000000
data:([] time:.z.P+n?0D00:00:00.000000000;
        sym:n?`AAPL`GOOG`MSFT`TSLA;
        price:100+n?10f;
        size:n?1000)

// Store data in-memory and measure ingestion speed
\t insert[`trade_data] data
```

# Demo 2

```q
// Generate 10 million trades
n:10000000
trade_data:([] time:.z.P+n?0D00:00:00.000; 
               sym:n?`AAPL`GOOG`MSFT`TSLA; 
               price:100+n?10f; 
               size:10+n?990)

// Check data
select from trade_data where sym=`AAPL

// Compute VWAP grouped by minute
\t results: select vwap: size wavg price by time.minute from trade_data
```
